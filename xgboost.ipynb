```python
# --- Step 0: imports, paths, logging ---
# Crash-resilience plan: every expensive thing (raw events, per-market feature
# tables) is cached to disk. Re-running any cell is cheap and skips done work.

import os, time, pickle, logging, random
from pathlib import Path
from collections import defaultdict

import pandas as pd
import numpy as np

from polycluster import (
    get_market_by_slug,
    get_market_orderfilled_events,
    parse_orderfilled_events,
    build_history_df_from_orderfilled,
    compute_market_user_features,
)

RANDOM_SEED = 42
random.seed(RANDOM_SEED); np.random.seed(RANDOM_SEED)

NEG_PER_POS = 6  # 6 non-insiders per insider, per market

CACHE_DIR     = Path("cache")
EVENTS_DIR    = CACHE_DIR / "events"
FEATURES_DIR  = CACHE_DIR / "features"
EVENTS_DIR.mkdir(parents=True, exist_ok=True)
FEATURES_DIR.mkdir(parents=True, exist_ok=True)

# Filesystem-safe slug (some slugs are >200 chars and would blow ecryptfs/etc)
def slug_key(slug: str) -> str:
    return slug if len(slug) <= 180 else f"{slug[:170]}__{abs(hash(slug)) % 10**8}"

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s  %(message)s",
    datefmt="%H:%M:%S",
    force=True,
)
log = logging.getLogger("xgb")
log.info(f"Setup OK. cache dir: {CACHE_DIR.resolve()}")
```

```python
# --- Step 1: group the 158 known pairs by market ---
# insiders_by_market[slug] -> list of insider wallets in that market
# all_insider_wallets       -> *global* exclusion set for negative sampling
#                              (a wallet that's an insider in market A should NOT
#                              be sampled as a "non-insider" in market B — that
#                              would inject label noise.)

insiders_by_market: dict[str, list[str]] = defaultdict(list)
for wallet, slug in KNOWN_INSIDER_PAIRS:
    insiders_by_market[slug].append(wallet.lower())

all_insider_wallets = {w.lower() for w, _ in KNOWN_INSIDER_PAIRS}
market_slugs = sorted(insiders_by_market.keys())

log.info(f"{len(market_slugs)} unique markets · {len(all_insider_wallets)} insider wallets")
log.info("Top 10 markets by insider count:")
for slug, ws in sorted(insiders_by_market.items(), key=lambda kv: -len(kv[1]))[:10]:
    log.info(f"  {len(ws):2d} insiders  ·  {slug[:90]}")
```

```python
# --- Step 2: fetch raw OrderFilled events per market, cached to disk ---
# This is the slow step (network + subgraph paging). Each market is pickled to
# cache/events/<slug>.pkl. Re-running the cell skips any already-cached market.
# We wrap each market in try/except so one failure doesn't abort the whole run.

def fetch_events_cached(slug: str):
    """Returns (market, events) or None on failure. Reads from disk if cached."""
    cache_path = EVENTS_DIR / f"{slug_key(slug)}.pkl"
    if cache_path.exists():
        with open(cache_path, "rb") as f:
            return pickle.load(f)
    market = get_market_by_slug(slug)
    result = get_market_orderfilled_events(market, checkpointed=True, verbose=False)
    events = result["rows"] if isinstance(result, dict) else result
    with open(cache_path, "wb") as f:
        pickle.dump((market, events), f)
    return (market, events)

fetch_failures: list[tuple[str, str]] = []
log.info(f"=== Step 2: fetching events for {len(market_slugs)} markets ===")

for i, slug in enumerate(market_slugs, 1):
    t0 = time.time()
    cache_path = EVENTS_DIR / f"{slug_key(slug)}.pkl"
    was_cached = cache_path.exists()
    try:
        market, events = fetch_events_cached(slug)
        log.info(
            f"[{i:2d}/{len(market_slugs)}] {'CACHED ' if was_cached else 'FETCHED'}  "
            f"{len(events):>7,} events  ·  {time.time()-t0:6.1f}s  ·  {slug[:60]}"
        )
    except Exception as e:
        fetch_failures.append((slug, f"{type(e).__name__}: {e}"))
        log.error(f"[{i:2d}/{len(market_slugs)}] FAILED  {slug[:60]}: {type(e).__name__}: {e}")

log.info(f"=== Step 2 done · {len(market_slugs)-len(fetch_failures)} ok · {len(fetch_failures)} failed ===")
for slug, msg in fetch_failures:
    log.warning(f"  fail: {slug[:80]} — {msg}")
```

```python
# --- Step 3: sample 6 non-insiders per insider, per market (volume-floor matched) ---
# Why this isn't uniform random:
#   Most wallets that touch a market are casual traders (tiny one-off bets).
#   Uniform sampling fills the negative class with casuals, so the model only
#   has to learn "serious whale vs casual" -- not "insider vs honest serious
#   trader." It would then flag legitimate big traders as insiders.
# Fix (hard-negative mining, light version):
#   For each market, compute each wallet's total USDC volume traded. Take the
#   25th percentile of the volume distribution *among that market's insiders*
#   as a floor. Only wallets at or above that floor are eligible negatives.
#   The model now has to distinguish insiders from comparably-serious traders.

labeled_rows: list[dict] = []
sampling_warnings: list[tuple[str, int, int]] = []  # (slug, wanted, got)
log.info(f"=== Step 3: sampling {NEG_PER_POS}:1 negatives per market (volume-floor matched) ===")

for i, slug in enumerate(market_slugs, 1):
    cache_path = EVENTS_DIR / f"{slug_key(slug)}.pkl"
    if not cache_path.exists():
        log.warning(f"[{i:2d}/{len(market_slugs)}] SKIP (no event cache): {slug[:60]}")
        continue
    with open(cache_path, "rb") as f:
        market, events = pickle.load(f)

    insiders = insiders_by_market[slug]
    try:
        parsed = parse_orderfilled_events(events, market.token_ids, market.outcomes)
    except Exception as e:
        log.error(f"[{i:2d}/{len(market_slugs)}] parse FAILED {slug[:60]}: {e}")
        continue

    # Per-wallet total USDC volume in this market (sum of cash legs over all trades).
    wallet_volume: dict[str, float] = {
        w.lower(): sum(float(t["cash_amount"]) for t in trades)
        for w, trades in parsed["wallet_trades"].items()
    }
    traded = set(wallet_volume.keys())

    insiders_present = [w for w in insiders if w in traded]
    missing = [w for w in insiders if w not in traded]
    if missing:
        log.warning(
            f"  {len(missing)} insider(s) absent from event log for {slug[:50]}: "
            f"{missing[:3]}{'…' if len(missing) > 3 else ''}"
        )

    n_pos = len(insiders_present)
    n_want = NEG_PER_POS * n_pos

    if n_pos == 0:
        log.warning(f"[{i:2d}/{len(market_slugs)}] no insiders present in events → skip {slug[:60]}")
        continue

    raw_pool = sorted(traded - all_insider_wallets)

    # Volume floor = 25th percentile of insider volumes for this market.
    insider_vols = np.array([wallet_volume[w] for w in insiders_present], dtype=float)
    vol_floor = float(np.percentile(insider_vols, 25))

    eligible_pool = [w for w in raw_pool if wallet_volume[w] >= vol_floor]

    if len(eligible_pool) < n_want:
        sampling_warnings.append((slug, n_want, len(eligible_pool)))
        sampled = eligible_pool
    else:
        rng = random.Random(f"{RANDOM_SEED}-{slug}")  # deterministic per-market
        sampled = rng.sample(eligible_pool, n_want)

    for w in insiders_present:
        labeled_rows.append({"wallet": w, "market_slug": slug, "is_insider": 1})
    for w in sampled:
        labeled_rows.append({"wallet": w, "market_slug": slug, "is_insider": 0})

    log.info(
        f"[{i:2d}/{len(market_slugs)}] {n_pos:2d} pos + {len(sampled):4d} neg  "
        f"(pool {len(raw_pool):>5d} → {len(eligible_pool):>5d} ≥ ${vol_floor:,.0f} floor)  ·  {slug[:50]}"
    )

labeled_df = pd.DataFrame(labeled_rows)
labeled_df.to_parquet(CACHE_DIR / "labeled_pairs.parquet", index=False)
n_pos, n_neg = int(labeled_df.is_insider.sum()), int((1 - labeled_df.is_insider).sum())
log.info(f"=== Step 3 done · {len(labeled_df)} labeled rows · {n_pos} pos / {n_neg} neg ===")
log.info(f"Saved → cache/labeled_pairs.parquet")
if sampling_warnings:
    log.warning(f"{len(sampling_warnings)} market(s) had insufficient negatives after volume filter:")
    for s, w, g in sampling_warnings:
        log.warning(f"  wanted {w:3d}, got {g:3d}: {s[:60]}")
```

```python
# --- Step 4: compute features per (wallet, market), cached per market ---
# We bypass build_feature_dataframe and use the lower-level pieces directly so
# we can reuse the cached events from Step 2 (no second network round-trip).

def features_for_market(slug: str, wallets: list[str]) -> pd.DataFrame | None:
    """Compute feature rows for the given wallets in one market.
    Returns a DataFrame (one row per wallet) or None on hard failure."""
    feat_path = FEATURES_DIR / f"{slug_key(slug)}.parquet"
    if feat_path.exists():
        return pd.read_parquet(feat_path)

    cache_path = EVENTS_DIR / f"{slug_key(slug)}.pkl"
    if not cache_path.exists():
        log.error(f"  no event cache for {slug[:50]}")
        return None

    with open(cache_path, "rb") as f:
        market, events = pickle.load(f)

    if not events:
        log.warning(f"  empty event log for {slug[:50]} — skipping")
        return None

    # Auto-derive final_outcome_yes from gamma's outcome_prices on closed markets
    # (1 if YES settled >= 0.5, else 0). Open markets stay None and the
    # outcome-dependent features become NaN — that's fine.
    final_outcome_yes = None
    if market.closed and market.outcome_prices:
        final_outcome_yes = 1 if float(market.outcome_prices[0]) >= 0.5 else 0

    parsed = parse_orderfilled_events(events, market.token_ids, market.outcomes)
    history_df = build_history_df_from_orderfilled(
        events, yes_token=market.yes_token_id, no_token=market.no_token_id,
    )

    rows = []
    n_empty = 0
    n_err = 0
    for w in wallets:
        wl = w.lower()
        trades = parsed["wallet_trades"].get(wl, [])
        if not trades:
            rows.append({"wallet": wl, "market_slug": slug})
            n_empty += 1
            continue
        try:
            feats = compute_market_user_features(
                parsed_trades=trades,
                history_df=history_df,
                market_start_time=market.start_ts,
                market_end_time=market.end_ts,
                final_outcome_yes=final_outcome_yes,
            )
            feats["wallet"] = wl
            feats["market_slug"] = slug
            rows.append(feats)
        except Exception as e:
            n_err += 1
            log.warning(f"    feat fail wallet={wl[:10]}.. {type(e).__name__}: {e}")
            rows.append({"wallet": wl, "market_slug": slug})

    df = pd.DataFrame(rows)
    df.to_parquet(feat_path, index=False)
    if n_empty or n_err:
        log.info(f"    ({n_empty} empty trades · {n_err} feature errors)")
    return df

log.info(f"=== Step 4: computing features for {len(market_slugs)} markets ===")
feature_failures: list[str] = []
per_market_dfs: list[pd.DataFrame] = []

for i, slug in enumerate(market_slugs, 1):
    t0 = time.time()
    wallets = labeled_df.loc[labeled_df.market_slug == slug, "wallet"].tolist()
    if not wallets:
        log.warning(f"[{i:2d}/{len(market_slugs)}] no labeled wallets for {slug[:60]} — skip")
        continue
    feat_path = FEATURES_DIR / f"{slug_key(slug)}.parquet"
    was_cached = feat_path.exists()
    try:
        df = features_for_market(slug, wallets)
        if df is None:
            feature_failures.append(slug)
            continue
        per_market_dfs.append(df)
        log.info(
            f"[{i:2d}/{len(market_slugs)}] {'CACHED  ' if was_cached else 'COMPUTED'}  "
            f"{len(df):>4d} rows  ·  {time.time()-t0:6.1f}s  ·  {slug[:60]}"
        )
    except Exception as e:
        feature_failures.append(slug)
        log.error(f"[{i:2d}/{len(market_slugs)}] FAILED  {slug[:60]}: {type(e).__name__}: {e}")

log.info(f"=== Step 4 done · {len(per_market_dfs)} ok · {len(feature_failures)} failed ===")
for s in feature_failures:
    log.warning(f"  fail: {s[:80]}")
```

```python
# --- Step 5: concat per-market feature tables, attach labels, save ---
features_df = pd.concat(per_market_dfs, ignore_index=True)
dataset = features_df.merge(labeled_df, on=["wallet", "market_slug"], how="left")

# Every feature row should have come from a labeled pair — assert that
n_unlabeled = int(dataset["is_insider"].isna().sum())
if n_unlabeled:
    log.warning(f"{n_unlabeled} feature rows have no label (will be dropped)")
    dataset = dataset.dropna(subset=["is_insider"])
dataset["is_insider"] = dataset["is_insider"].astype(int)

out_path = CACHE_DIR / "training_data.parquet"
dataset.to_parquet(out_path, index=False)
log.info(f"=== Saved {len(dataset)} rows × {len(dataset.columns)} cols → {out_path} ===")
log.info(f"Class balance: {dict(dataset['is_insider'].value_counts())}")
log.info(f"Markets in dataset: {dataset['market_slug'].nunique()}")
log.info(f"Unique wallets in dataset: {dataset['wallet'].nunique()}")

# Quick missingness peek — useful for picking imputation strategy later
nulls = dataset.drop(columns=["wallet", "market_slug", "is_insider"]).isna().mean().sort_values(ascending=False)
print("\nTop 15 most-missing feature columns (fraction NaN):")
print(nulls.head(15).round(3).to_string())

dataset.head()
```

```python
# --- Step 6: save the labeled feature dataset ---
# Persists `dataset` to parquet so the model cells below can run independently
# of the slow data-build steps (0-5). Also drops a sidecar JSON describing the
# schema — feature columns, class counts, market/wallet counts — so we can
# spot silent schema drift between runs and so inference code knows the
# expected feature order.

import json

dataset_path = CACHE_DIR / "training_data.parquet"
schema_path  = CACHE_DIR / "training_data.schema.json"

dataset.to_parquet(dataset_path, index=False)

feature_cols = [c for c in dataset.columns if c not in {"wallet", "market_slug", "is_insider"}]
schema = {
    "n_rows": int(len(dataset)),
    "n_features": len(feature_cols),
    "feature_columns": feature_cols,
    "class_counts": {str(k): int(v) for k, v in dataset["is_insider"].value_counts().items()},
    "n_markets": int(dataset["market_slug"].nunique()),
    "n_wallets": int(dataset["wallet"].nunique()),
}
with open(schema_path, "w") as f:
    json.dump(schema, f, indent=2)

log.info(f"Saved dataset  → {dataset_path}  ({schema['n_rows']:,} rows × {schema['n_features']} features)")
log.info(f"Saved schema   → {schema_path}")
log.info(f"Class counts   → {schema['class_counts']}")
```

```python
# --- Step 7: train an XGBoost classifier (with class-imbalance handling) ---
# Class imbalance is the key wrinkle:
#   We have ~6 negatives per positive. A trivial "predict 0 for everyone"
#   classifier hits ~86% accuracy and catches zero insiders. Two levers:
#     1) scale_pos_weight = n_neg / n_pos. XGBoost multiplies the loss for
#        positive examples by this number, so missing a real insider hurts
#        ~6× as much as misclassifying a non-insider. This is THE standard
#        knob for imbalance in xgboost binary classification.
#     2) eval_metric="aucpr" (precision-recall AUC). With heavy imbalance,
#        ROC-AUC stays optimistic; PR-AUC focuses on the rare-class ranking.
# NaN handling:
#   XGBoost natively handles missing values — at each split it learns a
#   default direction for missing. So we leave NaN as-is (no imputation).
# Train/test split:
#   Stratified random row split (test_size=20%). A wallet may appear in BOTH
#   splits across different markets — mild leakage risk we accept for the
#   first cut. If results look too good (PR-AUC > 0.95), switch to a
#   per-wallet GroupKFold and retrain.

import xgboost as xgb
from sklearn.model_selection import train_test_split

ID_COLS = ["wallet", "market_slug"]
LABEL   = "is_insider"

y = dataset[LABEL].astype(int)
X = dataset.drop(columns=ID_COLS + [LABEL]).select_dtypes(include=["number"])
log.info(f"Feature matrix: {X.shape[0]:,} rows × {X.shape[1]} cols")

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=RANDOM_SEED,
)
log.info(f"Train: {len(X_train):,} rows ({int(y_train.sum())} pos)")
log.info(f"Test : {len(X_test):,} rows ({int(y_test.sum())} pos)")

# Imbalance weight — computed from TRAIN ONLY (never peek at test for params).
n_neg_train = int((y_train == 0).sum())
n_pos_train = int((y_train == 1).sum())
scale_pos_weight = n_neg_train / max(n_pos_train, 1)
log.info(f"scale_pos_weight = {n_neg_train}/{n_pos_train} = {scale_pos_weight:.2f}")

model = xgb.XGBClassifier(
    n_estimators=500,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,
    eval_metric="aucpr",
    tree_method="hist",
    random_state=RANDOM_SEED,
    n_jobs=-1,
)
model.fit(
    X_train, y_train,
    eval_set=[(X_train, y_train), (X_test, y_test)],
    verbose=50,
)
log.info("=== Step 7 done · model trained ===")
```

```python
# --- Step 8: evaluate the model ---
# Why we don't trust accuracy here:
#   With ~6 negatives per positive, always predicting "non-insider" hits ~86%
#   accuracy and catches zero insiders. Useless. So we use:
#   - PR-AUC  (average precision): area under precision-recall. Best metric
#     for rare-class problems — sensitive to how well the model ranks the few
#     real positives at the top. Chance level = positive rate (~0.14 here).
#   - ROC-AUC: probability a random insider gets a higher score than a random
#     non-insider. Less sensitive to imbalance but easy to interpret.
#   - Precision @ 0.5: of wallets we flagged, what fraction are real insiders.
#   - Recall @ 0.5: of real insiders, what fraction we caught.
#   - Confusion matrix to see where errors fall (false positives vs false negs).

from sklearn.metrics import (
    classification_report,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
)

y_proba_test = model.predict_proba(X_test)[:, 1]
y_pred_test  = (y_proba_test >= 0.5).astype(int)

roc_auc = roc_auc_score(y_test, y_proba_test)
pr_auc  = average_precision_score(y_test, y_proba_test)

print(f"ROC-AUC: {roc_auc:.4f}")
print(f"PR-AUC : {pr_auc:.4f}   (chance level = {y_test.mean():.4f})")
print()
print("Classification report @ threshold 0.5:")
print(classification_report(
    y_test, y_pred_test,
    target_names=["non-insider", "insider"],
    digits=3,
))
print("Confusion matrix  [rows = true, cols = pred]:")
print(pd.DataFrame(
    confusion_matrix(y_test, y_pred_test),
    index=["true non-insider", "true insider"],
    columns=["pred non-insider", "pred insider"],
))

print("\nTop 20 features by importance:")
imp = pd.Series(model.feature_importances_, index=X.columns).sort_values(ascending=False)
print(imp.head(20).round(4).to_string())
```

```python
# --- Step 9: save the trained model ---
# Two artifacts:
#   - xgb_insider_<timestamp>.json  → the model in xgboost's portable JSON
#     format (preferred over pickle — stable across xgboost versions).
#   - xgb_insider_<timestamp>.meta.json → feature column order + sizes +
#     metrics + scale_pos_weight, so inference code can rebuild the feature
#     matrix in the exact same column order and you can audit which model
#     produced which numbers.
# Also writes "xgb_insider_latest.*" pointers so downstream code can grab the
# most recent model without knowing the timestamp.

import json
from datetime import datetime

MODEL_DIR = CACHE_DIR / "models"
MODEL_DIR.mkdir(parents=True, exist_ok=True)
stamp = datetime.now().strftime("%Y%m%d_%H%M%S")

model_path = MODEL_DIR / f"xgb_insider_{stamp}.json"
meta_path  = MODEL_DIR / f"xgb_insider_{stamp}.meta.json"

model.save_model(str(model_path))
meta = {
    "trained_at": stamp,
    "n_train": int(len(X_train)),
    "n_test": int(len(X_test)),
    "n_features": int(X.shape[1]),
    "features": list(X.columns),
    "scale_pos_weight": float(scale_pos_weight),
    "metrics": {
        "roc_auc": float(roc_auc),
        "pr_auc":  float(pr_auc),
    },
}
with open(meta_path, "w") as f:
    json.dump(meta, f, indent=2)

# Latest pointers — overwritten each run.
model.save_model(str(MODEL_DIR / "xgb_insider_latest.json"))
with open(MODEL_DIR / "xgb_insider_latest.meta.json", "w") as f:
    json.dump(meta, f, indent=2)

log.info(f"Saved model → {model_path}")
log.info(f"Saved meta  → {meta_path}")
log.info(f"Latest pointer → {MODEL_DIR / 'xgb_insider_latest.json'}")
```
